```python
#14

a, b, c, d = map(int, input().split())

if not all(0 <= x <= 100 for x in [a, b, c, d]):
    print('잘못된 점수')
elif (a + b + c + d) // 4 >= 80:
    print('합격')
else:
    print('불합격')
```
